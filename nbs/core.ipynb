# cl_fit_vae_on_signals

`sample_dataset_parameters` is a convenience function yielding the input files for a *sample* dataset

```python
input_files = [e.as_posix() for e in trapallada.data.sample_dataset_parameters(include_parameters=True).values()]
input_files
```

Some labels might be available

```python
labels_files = sum([[prefix, e.as_posix()] for e, prefix in zip(trapallada.data.labels_files().values(), ['--reference-signals-file', '--annotated-windows-file'])], [])
labels_files
```

The final list of arguments also includes
- the *parameters file* at the beginning, and
- the *output directory* (optional) at the end.

```python
# arguments = [parameters_file.as_posix()] + input_files + ['-o', '/tmp']
arguments = [parameters_file.as_posix()] + input_files + labels_files + ['-o', '/tmp']
arguments
```

```python
trapallada.core.cl_fit_vae_on_signals(arguments)
```

# cl_grid_fit_vae_on_signals

A YAML parser

```python
yaml = YAML()
yaml.preserve_quotes = True
```

```python
with parameters_file.open('r') as f:
    
    parameters = yaml.load(f)
```

`parameters` is modified...

```python
trapallada.util.set_path_in_dict(('grid search', 'model.dim_z'), parameters, [2, 3])
trapallada.util.set_path_in_dict(('grid search', 'loss.beta'), parameters, [1, 2])
```

...and saved with a new name in a temporary directory

```python
new_parameters_file = pathlib.Path('/tmp') / parameters_file.name
with new_parameters_file.open('w') as f:
    yaml.dump(parameters, f)
new_parameters_file
```

```python
trapallada.core.cl_grid_fit_vae_on_signals([new_parameters_file.as_posix()] + arguments[1:-2])
```
