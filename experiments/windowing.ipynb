A plot with **all** the windows for **all** the signals, colored by the latter.

```python
fig = px.scatter(analysis_df.reset_index(), x='tsne_1', y='tsne_2', color='signal', hover_data=on_plot_hover_columns, opacity=0.5, width=1_200, height=800)
fig.show()
```
